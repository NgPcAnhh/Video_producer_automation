## 🎙️ BƯỚC 4: TẠO GIỌNG ĐỌC AI (EDGE-TTS) & ĐỒNG BỘ TIMING ÂM THANH
### 🎯 Mục đích & Chức năng:
- **Tạo Thoại AI (TTS)**: Đọc thoại tiếng Anh cho từng phân cảnh sử dụng giọng `en-GB-RyanNeural`.
- **Đồng bộ Timing**: Tự động đo thời lượng chính xác của từng câu thoại (giây), thêm 0.25s ngắt nghỉ.
- **Xuất dữ liệu**: Xuất file âm thanh tổng `full_narration_audio.mp3` và lưu metadata `scenes_timed.json` chứa `start_time`, `end_time`, `duration_seconds` của từng cảnh.

[⬆️ Trở về Mục lục](#table-of-contents)


```python
import asyncio
import edge_tts
import nest_asyncio
from pydub import AudioSegment
from IPython.display import Audio, display
import os
import tempfile
import shutil
import json

# Cho phép chạy asyncio loop trong Google Colab
nest_asyncio.apply()

# 1. Sử dụng thư mục lưu trữ đã khai báo tập trung ở trên
os.makedirs(audio_dir, exist_ok=True)
temp_audio_dir = tempfile.mkdtemp()

# Chọn Giọng Đọc (Microsoft Neural Voice):
VOICE_NAME = "en-US-AnaNeural"
PAUSE_BETWEEN_SCENES_MS = 0   # 0.25 giây khoảng lặng giữa các cảnh

# 2. Phân tích chuỗi JSON từ biến scenes_data ở cell phía trên
parsed_scenes = json.loads(scenes_data)

# 3. Hàm bất đồng bộ đọc script_en bằng Edge TTS
async def generate_scene_speech(text, output_file, voice):
    communicate = edge_tts.Communicate(text, voice)
    await communicate.save(output_file)

combined_audio = AudioSegment.empty()
total_scenes = len(parsed_scenes)
print(f"🎙️ Bắt đầu đọc audio và TỰ ĐỘNG TÍNH THỜI LƯỢNG (duration) từ script_en cho {total_scenes} Cảnh...")
print(f"🔊 Giọng đọc: [{VOICE_NAME}]")
print("=" * 60)

current_time_offset = 0.0

try:
    for idx, item in enumerate(parsed_scenes, 1):
        scene_num = item.get("scene_number", idx)
        title = item.get("title", f"Scene_{scene_num}")

        # Trích xuất script_en
        script_text = item.get("script_en") or item.get("script") or item.get("narration") or ""
        script_text = script_text.strip()

        filename = f"scene_{scene_num:02d}_temp_audio.mp3"
        temp_audio_path = os.path.join(temp_audio_dir, filename)
        print(f"[{idx}/{total_scenes}] 🔊 Scene {scene_num}: {title}")

        if not script_text:
            print("   ⚠️ Không tìm thấy script_en, gán mặc định duration_seconds = 5.0s.")
            start_t = round(current_time_offset, 2)
            dur = 5.0
            end_t = round(start_t + dur, 2)
            current_time_offset += dur
            item["start_time"] = start_t
            item["end_time"] = end_t
            item["duration_seconds"] = dur
            continue

        preview_text = script_text[:100]
        print(f"   📖 Script_EN: '{preview_text}...'")

        # Đọc thoại và tạo file audio MP3 tạm thời bằng Edge TTS
        try:
            asyncio.run(generate_scene_speech(script_text, temp_audio_path, VOICE_NAME))
        except Exception:
            loop = asyncio.get_event_loop()
            loop.run_until_complete(generate_scene_speech(script_text, temp_audio_path, VOICE_NAME))

        # Đọc file audio vừa tạo để lấy độ dài chính xác (giây)
        scene_audio = AudioSegment.from_file(temp_audio_path)
        raw_speech_duration = len(scene_audio) / 1000.0

        # Nối vào file tổng + thêm 0.25s khoảng lặng giữa các cảnh
        silent_pause = AudioSegment.silent(duration=PAUSE_BETWEEN_SCENES_MS)
        combined_audio += scene_audio + silent_pause

        # TỰ ĐỘNG TÍNH & GÁN TIMING CHÍNH XÁC CHO SCENE (start, end, duration)
        start_t = round(current_time_offset, 2)
        calculated_duration = round(raw_speech_duration + (PAUSE_BETWEEN_SCENES_MS / 1000.0), 2)
        end_t = round(start_t + calculated_duration, 2)
        current_time_offset += calculated_duration

        item["start_time"] = start_t
        item["end_time"] = end_t
        item["duration_seconds"] = calculated_duration

        print(f"   ⏱️ Thoại: {round(raw_speech_duration, 2)}s | Trống: {PAUSE_BETWEEN_SCENES_MS/1000.0}s | Khung thời gian: {start_t}s ➔ {end_t}s (Tổng: {calculated_duration}s)")
        print(f"   ✅ Đã ghép audio scene {scene_num} vào file tổng")
        print("-" * 50)

    # 4. Xuất FILE AUDIO DUY NHẤT & Metadata thời gian chuẩn xác
    merged_audio_path = os.path.join(audio_dir, "full_narration_audio.mp3")
    combined_audio.export(merged_audio_path, format="mp3")

    timed_metadata_path = os.path.join(audio_dir, "scenes_timed.json")
    with open(timed_metadata_path, "w", encoding="utf-8") as f:
        json.dump(parsed_scenes, f, ensure_ascii=False, indent=2)

    print("=" * 60)
    print(f"🎉 HOÀN THÀNH TẤT CẢ! Đã tự động tính xong thời lượng chính xác cho toàn bộ {total_scenes} cảnh.")
    print(f"➡️ File audio duy nhất đã lưu tại: {merged_audio_path}")
    print(f"➡️ Metadata phân cảnh khớp thời gian đã lưu tại: {timed_metadata_path}")
    print(f"📊 Tổng thời lượng audio: {round(len(combined_audio) / 1000, 2)} giây")
    display(Audio(merged_audio_path))

finally:
    # Dọn dẹp các mảnh ghép audio tạm thời
    if os.path.exists(temp_audio_dir):
        shutil.rmtree(temp_audio_dir)
        print("<ctrl42> Đã dọn dẹp các mảnh ghép audio tạm thời!")
```

## 🖼️ BƯỚC 5: TRÍCH XUẤT & CHUẨN HÓA HÌNH ẢNH TỪ FILE PDF (PYMUPDF)
### 🎯 Mục đích & Chức năng:
- **Trích xuất từ PDF**: Tự động đọc các file PDF thiết kế hình ảnh (`clean_1.pdf`, `clean_2.pdf`...).
- **Xử lý Hệ màu**: Chuyển đổi định dạng CMYK/PNG sang chuẩn màu RGB JPEG chất lượng cao.
- **Đánh số thứ tự**: Lưu trữ ảnh đánh số nối tiếp vào thư mục `clean_image` (`clean_001.jpg`, `clean_002.jpg`...).

[⬆️ Trở về Mục lục](#table-of-contents)


```python
import fitz  # PyMuPDF
import os

# Đường dẫn mặc định (giả định biến save_dir, drive_base đã có, nếu chưa bạn có thể gán cứng đường dẫn vào đây)
# Sử dụng clean_image_dir tập trung đã khai báo ở Bước 3
os.makedirs(clean_image_dir, exist_ok=True)
os.makedirs(clean_image_dir, exist_ok=True) # Đảm bảo thư mục lưu ảnh mới tồn tại

global_img_idx = 1 # Biến đếm thứ tự ảnh nối tiếp toàn bộ dự án (1, 2, 3...)
pdf_num = 1        # Số thứ tự của file PDF (clean_1, clean_2...)

print("📖 Bắt đầu quét và trích xuất tối đa 20 ảnh đầu tiên từ mỗi file PDF...")

while True:
    # Các đường dẫn có thể chứa file PDF
    possible_paths = [
        os.path.join(save_dir, "image", f"clean_{pdf_num}.pdf") if 'save_dir' in locals() else "",
        os.path.join(save_dir, f"clean_{pdf_num}.pdf") if 'save_dir' in locals() else "",
        os.path.join(drive_base, "image", f"clean_{pdf_num}.pdf") if 'drive_base' in locals() else "",
        os.path.join(drive_base, f"clean_{pdf_num}.pdf") if 'drive_base' in locals() else "",
        os.path.join(save_dir, "image", f"clean_{pdf_num:02d}.pdf") if 'save_dir' in locals() else "",
        os.path.join(save_dir, f"clean_{pdf_num:02d}.pdf") if 'save_dir' in locals() else "",
        f"image/clean_{pdf_num}.pdf",
        f"clean_{pdf_num}.pdf"
    ]

    pdf_path = None
    for p in possible_paths:
        if p and os.path.exists(p):
            pdf_path = p
            break

    # Dừng nếu không tìm thấy file PDF
    if not pdf_path:
        if pdf_num == 1:
            print(f"⚠️ Không tìm thấy file clean_1.pdf để bắt đầu!")
        else:
            print(f"✅ Đã quét xong tất cả {pdf_num - 1} file PDF liên tục.")
        break

    print(f"\n📄 [PDF #{pdf_num}] Đang xử lý file: {pdf_path}")
    doc = fitz.open(pdf_path)

    # -------------------------------------------------------------------------
    # ĐIỀU KIỆN 3: CHỈ QUÉT VÀ LẤY 20 ẢNH ĐẦU TIÊN
    # -------------------------------------------------------------------------
    unique_xrefs = []
    for page in doc:
        for img in page.get_images(full=True):
            xref = img[0]
            if xref not in unique_xrefs:
                if len(unique_xrefs) < pdf_max_img:
                    unique_xrefs.append(xref)
                else:
                    # Nếu đã đạt 20 ảnh, thoát vòng lặp ảnh trong trang này
                    break

        # Nếu đã đạt 20 ảnh, thoát vòng lặp trang (dừng quét hoàn toàn)
        if len(unique_xrefs) >= 20:
            print("   ⚠️ Đã đạt giới hạn 20 ảnh. Các ảnh từ 21 trở đi bị loại bỏ (không trích xuất).")
            break

    print(f"   🔍 Trích xuất {len(unique_xrefs)} ảnh gốc từ file clean_{pdf_num}.pdf.")
    start_idx_for_pdf = global_img_idx

    # Thực hiện lưu ảnh
    for idx, xref in enumerate(unique_xrefs):
        base_image = doc.extract_image(xref)
        image_bytes = base_image["image"]
        raw_ext = base_image["ext"]
        ext = "jpg" if raw_ext == "jpeg" else raw_ext

        # -------------------------------------------------------------------------
        # ĐIỀU KIỆN 2: Tên file chỉ là số (VD: 1.jpg, 2.jpg)
        # -------------------------------------------------------------------------
        filename = f"{global_img_idx}.{ext}"

        # -------------------------------------------------------------------------
        # ĐIỀU KIỆN 1: Lưu vào folder clean_image
        # -------------------------------------------------------------------------
        output_file_path = os.path.join(clean_image_dir, filename)

        with open(output_file_path, "wb") as f:
            f.write(image_bytes)

        print(f"   ✅ Saved {filename}")
        global_img_idx += 1

    end_idx_for_pdf = global_img_idx - 1
    if start_idx_for_pdf <= end_idx_for_pdf:
        print(f"📌 Đã hoàn tất clean_{pdf_num}.pdf ➔ Đánh số ảnh từ {start_idx_for_pdf} đến {end_idx_for_pdf}")
    else:
        print(f"📌 Đã hoàn tất clean_{pdf_num}.pdf ➔ Không có ảnh nào được trích xuất.")

    # Sang file tiếp theo
    pdf_num += 1
    doc.close()

print(f"\n🎉 HOÀN THÀNH! Đã xử lý {pdf_num - 1} file PDF, trích xuất tổng cộng {global_img_idx - 1} ảnh (1 -> {global_img_idx - 1}) vào thư mục: {clean_image_dir}")
```

## 🎥 BƯỚC 7: KHỞI TẠO HIỆU ỨNG CHUYỂN ĐỘNG CAMERA 2.5D (KEN BURNS SIÊU TỐC)
### 🎯 Mục đích & Chức năng:
- **Tương thích Toàn diện**: Hỗ trợ đồng thời phiên bản MoviePy v1.x và v2.x.
- **Tối ưu Tốc độ Đột phá (OpenCV C++ Fast Engine)**:
  - Khắc phục triệt để tình trạng chậm của MoviePy cũ (giảm thời gian tạo frame từ ~620ms xuống **~1ms/frame**).
  - Tự động chuẩn hóa ảnh về khung hình chuẩn 16:9 Full HD (1920x1080) với vùng đệm 15% (`2208x1242`).
  - Áp dụng các hiệu ứng chuyển động ống kính mượt mà: `zoom_in`, `zoom_out`, `pan_right`, `pan_left`, `pan_up`.
- **Hiệu ứng Chuyển cảnh**: Cấu hình sẵn các hàm Crossfade, Fade In, Fade Out màn hình đen.

[⬆️ Trở về Mục lục](#table-of-contents)

```python
import os
import random
import cv2
import numpy as np
from PIL import Image

try:
    from moviepy.editor import VideoClip, ImageClip, AudioFileClip, concatenate_videoclips, CompositeVideoClip, vfx
    MOVIEPY_V1 = True
except ImportError:
    from moviepy import VideoClip, ImageClip, AudioFileClip, concatenate_videoclips, CompositeVideoClip, vfx
    MOVIEPY_V1 = False

# -------------------------------------------------------------------------
# 1. HÀM CHUẨN BỊ CANVAS ẢNH CỐT LÕI (PRE-SCALE & CENTER CROP ONCE)
# -------------------------------------------------------------------------
def prepare_base_canvas(image_path, target_w=1920, target_h=1080, margin=0.15):
    """
    Đọc ảnh và chuẩn hóa kích thước thành Canvas 16:9 với viền mở rộng (margin).
    Giúp hiệu ứng zoom/pan mượt mà không bị méo hình và chỉ cần tính toán 1 lần duy nhất!
    """
    w_base = int(target_w * (1 + margin))  # 2208
    h_base = int(target_h * (1 + margin))  # 1242

    # Đọc ảnh (Hỗ trợ cả Unicode path)
    img = None
    try:
        img = cv2.imread(image_path)
        if img is not None:
            img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
    except Exception:
        img = None

    if img is None:
        pil_img = Image.open(image_path).convert("RGB")
        img = np.array(pil_img)

    h_orig, w_orig = img.shape[:2]
    # Tính tỉ lệ phóng đại để bao phủ toàn bộ Canvas chuẩn
    scale = max(w_base / w_orig, h_base / h_orig)
    new_w = int(round(w_orig * scale))
    new_h = int(round(h_orig * scale))

    interp = cv2.INTER_AREA if scale < 1.0 else cv2.INTER_LINEAR
    resized = cv2.resize(img, (new_w, new_h), interpolation=interp)

    # Cắt chính giữa (Center Crop) lấy đúng kích thước (w_base, h_base)
    start_x = max(0, (new_w - w_base) // 2)
    start_y = max(0, (new_h - h_base) // 2)
    base_img = resized[start_y:start_y + h_base, start_x:start_x + w_base]

    # Đảm bảo kích thước chính xác tuyệt đối
    if base_img.shape[0] != h_base or base_img.shape[1] != w_base:
        base_img = cv2.resize(base_img, (w_base, h_base), interpolation=cv2.INTER_LINEAR)

    return base_img, w_base, h_base

# -------------------------------------------------------------------------
# 2. HÀM TẠO CHUYỂN ĐỘNG KEN BURNS SIÊU TỐC (~1ms/frame bằng C++ OpenCV)
# -------------------------------------------------------------------------
def create_ken_burns_motion(image_path, duration, motion_type="zoom_in", target_resolution=(1920, 1080)):
    """
    Tạo clip chuyển động Ken Burns siêu tốc với chuẩn Full HD 1920x1080.
    Tốc độ nhanh gấp >400 lần so với MoviePy gốc nhờ kỹ thuật Viewport Slicing.
    """
    target_w, target_h = target_resolution
    margin = 0.15
    duration = max(0.1, duration)

    # Nạp và chuẩn hóa ảnh nền 1 lần duy nhất
    base_img, w_base, h_base = prepare_base_canvas(image_path, target_w, target_h, margin)

    def make_frame(t):
        p = min(max(t / duration, 0.0), 1.0)

        if motion_type == "zoom_in":
            # Zoom từ toàn cảnh vào sâu 15%
            wv = int(w_base - (w_base - target_w) * p)
            hv = int(h_base - (h_base - target_h) * p)
            x1 = (w_base - wv) // 2
            y1 = (h_base - hv) // 2
            crop = base_img[y1:y1 + hv, x1:x1 + wv]
            return cv2.resize(crop, (target_w, target_h), interpolation=cv2.INTER_LINEAR)

        elif motion_type == "zoom_out":
            # Zoom từ cận cảnh ra toàn cảnh
            wv = int(target_w + (w_base - target_w) * p)
            hv = int(target_h + (h_base - target_h) * p)
            x1 = (w_base - wv) // 2
            y1 = (h_base - hv) // 2
            crop = base_img[y1:y1 + hv, x1:x1 + wv]
            return cv2.resize(crop, (target_w, target_h), interpolation=cv2.INTER_LINEAR)

        elif motion_type == "pan_right":
            # Lia máy ngang từ trái sang phải (chỉ cắt ma trận, không cần resize -> 0ms!)
            dx = w_base - target_w
            x1 = int(dx * p)
            y1 = (h_base - target_h) // 2
            return base_img[y1:y1 + target_h, x1:x1 + target_w]

        elif motion_type == "pan_left":
            # Lia máy ngang từ phải sang trái
            dx = w_base - target_w
            x1 = int(dx * (1.0 - p))
            y1 = (h_base - target_h) // 2
            return base_img[y1:y1 + target_h, x1:x1 + target_w]

        elif motion_type == "pan_up":
            # Lia máy dọc từ dưới lên trên
            dy = h_base - target_h
            x1 = (w_base - target_w) // 2
            y1 = int(dy * (1.0 - p))
            return base_img[y1:y1 + target_h, x1:x1 + target_w]

        else:
            # Mặc định: Zoom nhẹ
            wv = int(w_base - (w_base - target_w) * 0.5 * p)
            hv = int(h_base - (h_base - target_h) * 0.5 * p)
            x1 = (w_base - wv) // 2
            y1 = (h_base - hv) // 2
            crop = base_img[y1:y1 + hv, x1:x1 + wv]
            return cv2.resize(crop, (target_w, target_h), interpolation=cv2.INTER_LINEAR)

    clip = VideoClip(make_frame, duration=duration)
    if MOVIEPY_V1:
        clip.fps = 24
    else:
        clip = clip.with_fps(24)
    return clip

# -------------------------------------------------------------------------
# 3. HÀM ÁP DỤNG CROSSFADE, FADE IN, FADE OUT (TƯƠNG THÍCH CẢ MOVIEPY V1 & V2)
# -------------------------------------------------------------------------
def apply_crossfade(clip, d=0.5):
    """Mờ đè ảnh mới lên ảnh cũ mà không thông qua màn hình đen."""
    if MOVIEPY_V1:
        return clip.crossfadein(d)
    else:
        return clip.with_effects([vfx.CrossFadeIn(d)])

def apply_fadein(clip, d=1.0):
    """Làm mờ từ màn hình đen vào clip."""
    if MOVIEPY_V1:
        return clip.fadein(d)
    else:
        if hasattr(vfx, 'FadeIn'):
            return clip.with_effects([vfx.FadeIn(d)])
        elif hasattr(clip, 'fadein'):
            return clip.fadein(d)
        else:
            return clip

def apply_fadeout(clip, d=1.0):
    """Làm mờ clip sang màn hình đen."""
    if MOVIEPY_V1:
        return clip.fadeout(d)
    else:
        if hasattr(vfx, 'FadeOut'):
            return clip.with_effects([vfx.FadeOut(d)])
        elif hasattr(clip, 'fadeout'):
            return clip.fadeout(d)
        else:
            return clip

MOTIONS_LIST = ["zoom_in", "zoom_out", "pan_right", "pan_left", "pan_up"]
TRANSITION_DURATION = 0.5  # Thời gian mờ chồng giữa 2 ảnh (0.5 giây)
print("🚀 ĐÃ KHỞI TẠO XONG BỘ HIỆU ỨNG KEN BURNS SIÊU TỐC (OPENCV FAST ENGINE)!")
```

```python
# =====================================================
# BƯỚC 8: DỰNG VIDEO HOÀN CHỈNH & RENDER TỐI ƯU ĐA LUỒNG CPU
# =====================================================
import os
import re
import json
import random
import subprocess

try:
    from moviepy.editor import ImageClip, VideoFileClip, concatenate_videoclips, AudioFileClip, CompositeVideoClip, AudioClip, vfx
    MOVIEPY_V1 = True
except ImportError:
    from moviepy import ImageClip, VideoFileClip, concatenate_videoclips, AudioFileClip, CompositeVideoClip, AudioClip, vfx
    MOVIEPY_V1 = False

TRANSITION_DURATION = 0.5  # Thời gian chuyển cảnh giữa các phân cảnh ảnh

# -----------------------------------------------------
# HÀM BỔ TRỢ: TÌM KIẾM TỆP TIN & THƯ MỤC THÔNG MINH
# -----------------------------------------------------
def find_file_smart(filename_list, search_roots):
    """Tìm tệp tin đầu tiên khớp trong danh sách các thư mục gốc."""
    for root in search_roots:
        if not root or not os.path.exists(root):
            continue
        for current_root, dirs, files in os.walk(root):
            for f in files:
                if f.lower() in [name.lower() for name in filename_list]:
                    return os.path.join(current_root, f).replace("\\", "/")
    return None

def find_dir_smart(dir_keywords, search_roots):
    """Tìm thư mục khớp từ khóa."""
    for root in search_roots:
        if not root or not os.path.exists(root):
            continue
        for current_root, dirs, files in os.walk(root):
            for d in dirs:
                if d.lower() in [k.lower() for k in dir_keywords]:
                    return os.path.join(current_root, d).replace("\\", "/")
    return None

# Các vị trí gốc tìm kiếm trên Colab / Drive / Máy cục bộ
SEARCH_ROOTS = [
    "/content/drive/MyDrive/AI VIDEO",
    "/content/drive/MyDrive",
    "/content",
    "d:/AI Video",
    "C:/AI VIDEO",
    os.getcwd()
]

if 'save_dir' in locals() and save_dir and os.path.exists(save_dir):
    SEARCH_ROOTS.insert(0, save_dir.replace("\\", "/"))
if 'drive_base' in locals() and drive_base and os.path.exists(drive_base):
    SEARCH_ROOTS.insert(0, drive_base.replace("\\", "/"))

# 1. TÌM FILE AUDIO TỔNG (full_narration_audio.mp3)
target_audio_path = None
if 'merged_audio_path' in locals() and merged_audio_path and os.path.exists(merged_audio_path):
    target_audio_path = merged_audio_path.replace("\\", "/")
elif 'merged_audio_path' in globals() and merged_audio_path and os.path.exists(merged_audio_path):
    target_audio_path = merged_audio_path.replace("\\", "/")

if not target_audio_path:
    target_audio_path = find_file_smart(["full_narration_audio.mp3"], SEARCH_ROOTS)

if not target_audio_path:
    for r in SEARCH_ROOTS:
        a_dir = find_dir_smart(["audio"], [r])
        if a_dir and os.path.exists(a_dir):
            mp3s = [os.path.join(a_dir, f).replace("\\", "/") for f in os.listdir(a_dir) if f.endswith(".mp3")]
            if mp3s:
                target_audio_path = mp3s[0]
                break

if not target_audio_path or not os.path.exists(target_audio_path):
    raise FileNotFoundError(
        "❌ KHÔNG TÌM THẤY FILE AUDIO `full_narration_audio.mp3`!\n"
        "👉 Nếu bạn dùng Google Colab: Vui lòng đảm bảo đã Mount Google Drive ở [Bước 3] và đã chạy tạo audio ở [Bước 4]."
    )

audio_clip = AudioFileClip(target_audio_path)
total_audio_duration = audio_clip.duration
print(f"🎵 Đã tìm thấy và nạp Audio tổng: {target_audio_path} ({round(total_audio_duration, 2)} giây)")

# 2. XÁC ĐỊNH THƯ MỤC ẢNH & THƯ MỤC XUẤT VIDEO
project_dir = save_dir if ('save_dir' in globals() and save_dir) else (os.path.dirname(os.path.dirname(target_audio_path)) if "audio" in target_audio_path else os.path.dirname(target_audio_path))
clean_img_dir = clean_image_dir if ('clean_image_dir' in globals() and clean_image_dir) else os.path.join(project_dir, "clean_image").replace("\\", "/")
image_dir_path = image_dir if ('image_dir' in globals() and image_dir) else os.path.join(project_dir, "image").replace("\\", "/")
video_dir_path = video_dir if ('video_dir' in globals() and video_dir) else os.path.join(project_dir, "video").replace("\\", "/")

if not os.path.exists(clean_img_dir):
    found_clean = find_dir_smart(["clean_image"], SEARCH_ROOTS)
    if found_clean:
        clean_img_dir = found_clean

os.makedirs(video_dir_path, exist_ok=True)
print(f"📖 Thư mục ảnh trích xuất PDF: {clean_img_dir}")
print(f"📁 Thư mục xuất video: {video_dir_path}")

# 3. NẠP TỰ ĐỘNG DỮ LIỆU PHÂN CẢNH (parsed_scenes)
loaded_parsed_scenes = None
if 'parsed_scenes' in locals() and parsed_scenes:
    loaded_parsed_scenes = parsed_scenes
elif 'parsed_scenes' in globals() and parsed_scenes:
    loaded_parsed_scenes = parsed_scenes

if not loaded_parsed_scenes:
    timed_json_path = find_file_smart(["scenes_timed.json"], SEARCH_ROOTS)
    if timed_json_path and os.path.exists(timed_json_path):
        try:
            with open(timed_json_path, 'r', encoding='utf-8') as f:
                loaded_parsed_scenes = json.load(f)
            print(f"📖 Đã tự động nạp {len(loaded_parsed_scenes)} phân cảnh từ file: {timed_json_path}")
        except Exception as e:
            print(f"⚠️ Lỗi đọc scenes_timed.json: {e}")

if not loaded_parsed_scenes:
    s_data = scenes_data if 'scenes_data' in locals() else (scenes_data if 'scenes_data' in globals() else None)
    if s_data:
        try:
            loaded_parsed_scenes = json.loads(s_data) if isinstance(s_data, str) else s_data
            avg_dur = round(total_audio_duration / max(1, len(loaded_parsed_scenes)), 2)
            for s_idx, sc in enumerate(loaded_parsed_scenes):
                sc['start_time'] = round(s_idx * avg_dur, 2)
                sc['duration_seconds'] = avg_dur
                sc['end_time'] = round((s_idx + 1) * avg_dur, 2)
            print(f"⚠️ Đã chia đều thời lượng {len(loaded_parsed_scenes)} cảnh từ scenes_data.")
        except Exception as e:
            print(f"⚠️ Lỗi xử lý scenes_data: {e}")

if not loaded_parsed_scenes:
    raise NameError(
        "❌ KHÔNG TÌM THẤY DỮ LIỆU PHÂN CẢNH `parsed_scenes`!\n"
        "👉 Vui lòng chạy [Bước 1] khai báo scenes_data hoặc [Bước 4/6] tạo file scenes_timed.json."
    )

parsed_scenes = loaded_parsed_scenes

# 4. DỰNG CÁC PHÂN CẢNH CHÍNH TRONG RAM (Main Scenes với Ken Burns & Crossfade)
total_scenes = len(parsed_scenes)
print(f"🎬 Khởi tạo {total_scenes} cảnh ảnh chính với hiệu ứng Ken Burns siêu tốc & Crossfade ({TRANSITION_DURATION}s)...")
print("=" * 60)

clips = []
last_valid_image_path = None

for idx, item in enumerate(parsed_scenes, 1):
    scene_num = item.get("scene_number", idx)
    title = item.get("title", f"Scene_{scene_num}")

    image_path = None

    # Tìm ảnh trong clean_image
    if os.path.exists(clean_img_dir):
        candidates = [
            f"{scene_num}.jpg", f"{scene_num}.png", f"{scene_num}.jpeg",
            f"{scene_num:02d}.jpg", f"{scene_num:02d}.png",
            f"{scene_num:03d}.jpg", f"{scene_num:03d}.png"
        ]
        for candidate in candidates:
            full_p = os.path.join(clean_img_dir, candidate)
            if os.path.exists(full_p):
                image_path = full_p
                break

    # Tìm ảnh trong image
    if not image_path and os.path.exists(image_dir_path):
        possible_names = [
            get_safe_filename(scene_num, title, ext="jpg") if 'get_safe_filename' in locals() else f"scene_{scene_num:02d}.jpg",
            get_safe_filename(scene_num, title, ext="png") if 'get_safe_filename' in locals() else f"scene_{scene_num:02d}.png",
            f"scene_{scene_num:02d}_Scene_{scene_num}.jpg",
            f"scene_{scene_num:02d}_Scene_{scene_num}.png",
            f"scene_{scene_num}_Scene_{scene_num}.jpg",
            f"scene_{scene_num}_Scene_{scene_num}.png",
        ]
        for fname in possible_names:
            full_p = os.path.join(image_dir_path, fname)
            if os.path.exists(full_p):
                image_path = full_p
                break

        if not image_path:
            available_images = os.listdir(image_dir_path)
            prefix_pattern = f"scene_{scene_num:02d}_"
            alt_prefix = f"scene_{scene_num}_"
            matching = [f for f in available_images if f.startswith(prefix_pattern) or f.startswith(alt_prefix)]
            if matching:
                image_path = os.path.join(image_dir_path, matching[0])

    if not image_path:
        if last_valid_image_path:
            image_path = last_valid_image_path
            print(f"   ⚠️ [Cảnh {idx}/{total_scenes}] Không thấy ảnh cho scene_{scene_num}. Dùng lại ảnh trước.")
        else:
            print(f"   ❌ [Cảnh {idx}/{total_scenes}] Không tìm thấy bất kỳ ảnh nào!")
            continue
    else:
        last_valid_image_path = image_path

    spoken_duration = item.get("duration_seconds", total_audio_duration / total_scenes)
    spoken_duration = max(0.1, spoken_duration)

    if idx > 1:
        clip_duration = spoken_duration + TRANSITION_DURATION
    else:
        clip_duration = spoken_duration

    chosen_motion = random.choice(MOTIONS_LIST) if 'MOTIONS_LIST' in locals() else "zoom_in"
    # Clip trả về đã ở chuẩn Full HD 1920x1080 cực kỳ mượt mà, không cần resize lại
    clip = create_ken_burns_motion(image_path, clip_duration, motion_type=chosen_motion, target_resolution=(1920, 1080))

    if idx > 1:
        clip = apply_crossfade(clip, d=TRANSITION_DURATION)

    clips.append(clip)
    start_t = item.get("start_time", "N/A")
    end_t = item.get("end_time", "N/A")
    print(f"   [Cảnh {idx}/{total_scenes}] 📸 {os.path.basename(image_path)} | ⏰ [{start_t}s ➔ {end_t}s] | ⏱️ {round(spoken_duration, 2)}s | 🎥 {chosen_motion} | 🪄 Crossfade")

if not clips:
    raise ValueError("❌ Không có ảnh hợp lệ!")

print("-" * 60)
print("⏳ Đã tạo xong toàn bộ các phân cảnh chính trong bộ nhớ RAM.")

main_scenes_clip = concatenate_videoclips(clips, padding=-TRANSITION_DURATION, method="compose")

if MOVIEPY_V1:
    main_scenes_clip = main_scenes_clip.set_audio(audio_clip)
    main_scenes_clip = main_scenes_clip.set_duration(total_audio_duration)
    main_scenes_clip.fps = 24
else:
    main_scenes_clip = main_scenes_clip.with_audio(audio_clip)
    main_scenes_clip = main_scenes_clip.with_duration(total_audio_duration)
    main_scenes_clip.fps = 24

# =========================================================================
# XUẤT VIDEO HOÀN CHỈNH BẰNG CPU ĐA LUỒNG TỐI ƯU (VERYFAST)
# =========================================================================
final_video = main_scenes_clip

if 'output_video_path' not in globals() or not output_video_path:
    output_video_path = os.path.join(video_dir, "COMPLETE_VIDEO.mp4").replace("\\", "/")
else:
    output_video_path = output_video_path.replace("\\", "/")

if 'temp_audio_file' not in globals() or not temp_audio_file:
    temp_audio_file = os.path.join(video_dir, "temp_audio_track.m4a").replace("\\", "/")
else:
    temp_audio_file = temp_audio_file.replace("\\", "/")

print("=" * 60)
print(f"🎬 BẮT ĐẦU XUẤT VIDEO TẠI: {output_video_path}")
print("=" * 60)

threads_count = max(2, os.cpu_count() or 2)
print(f"⚙️ [CHẾ ĐỘ CPU] Đang render video bằng CPU đa luồng (threads={threads_count}, preset=veryfast)...")

try:
    final_video.write_videofile(
        output_video_path,
        fps=24,
        codec="libx264",
        audio_codec="aac",
        temp_audiofile=temp_audio_file,
        remove_temp=True,
        preset="veryfast",          # Nhanh gấp 4-5 lần so với medium truyền thống
        ffmpeg_params=[
            "-pix_fmt", "yuv420p",
            "-movflags", "+faststart",
            "-crf", "23"
        ],
        threads=threads_count
    )
    print("=" * 60)
    print("🎉 HOÀN TẤT RENDER VIDEO THÀNH CÔNG 100%!")
    print(f"➡️ File video đã lưu an toàn tại: {output_video_path}")
    print("=" * 60)
except Exception as cpu_err:
    print(f"❌ [LỖI RENDER CPU]: {cpu_err}")
    raise cpu_err
```
